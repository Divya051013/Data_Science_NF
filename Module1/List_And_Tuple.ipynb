```python
nums = [10, 5, 20, 8, 15]

largest = max(nums)
smallest = min(nums)

print("Largest:", largest)
print("Smallest:", smallest)
```

```python
nums = [1, 2, 2, 3, 4, 4, 5]

unique = []
for num in nums:
    if num not in unique:
            unique.append(num)
print(unique)
```

```python
nums = [1, 2, 3, 4, 5]

reversed_list = nums[::-1]

print(reversed_list)
```

```python
nums = [1, 2, 3, 4, 5, 6]

even = 0
odd = 0

for num in nums:
    if num % 2 == 0:
            even += 1
    else:
             odd += 1

print("Even:", even)
print("Odd:", odd)
```

```python
list1 = [3, 1, 4]
list2 = [6, 2, 5]

merged = list1 + list2
merged.sort()

print(merged)
```

```python
nums = [10, 20, 4, 45, 99]

first = second = float('-inf')

for num in nums:
    if num > first:
            second = first
            first = num
    elif num > second and num != first:
            second = num

print("Second Largest:", second)
```

```python
t = (1, 2, 3, 4, 5)

element = 3

if element in t:
    print("Exists")
else:
        print("Does not exist")
```

```python
t = (1, 2, 2, 3, 2, 4)

element = 2

print("Count:", t.count(element))
```

```python
t = [(1, 3), (4, 1), (2, 2)]

sorted_list = sorted(t, key=lambda x: x[1])

print(sorted_list)
```

```python
t = (1, 2, 3)

# Tuple to List
l = list(t)

# List to Tuple
t2 = tuple(l)

print("List:", l)
print("Tuple:", t2)
```
